## Fill The Null Value

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('/content/drive/MyDrive/only.data/Twitter Stock Market Dataset.csv')
df["Open"].fillna(df["Open"].mean(),inplace=True)
df["High"].fillna(df["High"].mean(),inplace=True)
df["Low"].fillna(df["Low"].mean(),inplace=True)
df["Close"].fillna(df["Close"].mean(),inplace=True)
df["Adj Close"].fillna(df["Adj Close"].mean(),inplace=True)
df["Volume"].fillna(df["Volume"].median(),inplace=True)
```

```python
df
```

split date

```python
df[["day","month","year"]]=df["Date"].str.split("-",expand=True)
df.head(23)
tt=df[["Close","Open"]]
tt.head(23)
```

```python
dd=df[df["year"]=="2021"]
sns.relplot(x='Close',y='Volume',data=dd,kind="line")
plt.grid(True)
plt.show()
dd.groupby("month").agg({"Close":["mean","min","max"]}).plot.bar()
plt.title("mean ,max , mean in 2021")
plt.show()
```

```python
#in day
df['Daily Change'] =df['Close'].diff()
plt.figure(figsize=(14, 7))
sns.barplot(x='day',y='Daily Change',data=df)
plt.grid(True)
plt.title("Twitter Stock Closing Prices Over Time")
plt.show()

#in day
df['Daily Change'] =df['Open'].diff()
plt.figure(figsize=(14, 7))
sns.barplot(x='day',y='Daily Change',data=df)
plt.grid(True)
plt.title("Twitter Stock Closing Prices Over Time")
plt.show()

#in month
df['Daily Change'] =df['Close'].diff()
plt.figure(figsize=(14, 7))
sns.barplot(x='month',y='Daily Change',data=df)
plt.grid(True)
plt.title("Twitter Stock Closing Prices Over Time")
plt.show()

#in month
df['Daily Change'] =df['Open'].diff()
plt.figure(figsize=(14, 7))
sns.barplot(x='month',y='Daily Change',data=df)
plt.grid(True)
plt.title("Twitter Stock Closing Prices Over Time")
plt.show()

#relation of open of every month and closinf of every month
sns.relplot(x='Close',y='Open',data=df)




```
